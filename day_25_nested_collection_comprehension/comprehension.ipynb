##### Comprehension
*easy way creating list|set|dictionary from a sequence*  

```
lst =[1,2,3,4,5]  

add_five =[]

for num in lst:

    result = num + 5

    add_five.append(result)



```

syntax:  

```
result = [return_value  iteration conditon]

```
EG:  

```
    lst =[1,2,3,4,5]  
    
    add_five = [num+5 for num in lst]

    print(add_five)

```

````python
lst =[10,11,12,13,14]

squares = [num**2 for num in lst]

print(squares)
````

````python
cube = [num**3 for num in lst]

print(cube)
````

````python
double = [num*2 for num in lst]
print(double)
````

````python
#greater than 13
lst =[10,11,12,13,14]
num_gt_13 = [num for num in lst if num>13]

print(num_gt_13)
````

````python
#even numbers 

evens=[num for num in lst if num%2==0]

print(evens)
````

````python

words=["flight","flower","fly","fluid"]

upper_words = [w.upper() for w in words] 

print(upper_words)


````

````python
# return words whose length > 3
words=["flight","flower","fly","fluid"]

word_filter = {w for w in words if len(w) > 3}

print(word_filter)
````

````python


lst=[10,11,12,11,10,10,11,13,14]

#{10:100,11:121,12:144}

squares={num:num**2 for num in lst}

print(squares)
````

````python
words=["forecast","foreword","foremost","forehead"]

words.sort()

first = words[0]

last = words[-1]

small_word = first if len(first) < len(last) else last

prefix = ""

for i in range(len(small_word)):

    if first[i] == last[i]:

        prefix+=first[i]
    else:

        break
print(prefix)
````
