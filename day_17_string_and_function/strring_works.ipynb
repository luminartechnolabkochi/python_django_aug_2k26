#### python provide string methods to handle string objects
**case conversion methods**

`lower()` => convert string object to lower case  
`upper()`=> convert string object to uppercase  

`is_digit()` => return true if string object is a digit return false otherwise  

`is_alpha()` => return True if string object contain only alphabets return false otherwise 



````python
company_name = "luminar technolab"

capitalized_company_name=company_name.upper()

print(capitalized_company_name)
print(company_name)
````

````python
employee_name="RAHUL VS"

lower_case_name=employee_name.lower()

print(lower_case_name)
````

````python
vehicle_number="kl-08-cd-1919"

is_numbers= vehicle_number.isdigit()

print(is_numbers)
````

````python

challan_number = "wertyuicvbnm"

is_alphabets_only = challan_number.isalpha()

print(is_alphabets_only)
````

````python

aadhar_number = "55511333"

print(aadhar_number.isdigit())
````

````python
pan_card_number ="DTZPS6463J"

print(pan_card_number.isalnum())
````
