```python
daily_calories =[1800,1700,1800,2000,1850,2100,2200]
#                  0    1    2    3    4    5    6

for i in range(0,len(daily_calories)):

    print(daily_calories[i])
```

```python
for cal in daily_calories:

    print(cal)
```

```python
#update list

daily_calories[0]=1900

print(daily_calories)
```

### list methods
- `append(value)`  => appends value to end of the list  
- `insert(pos,value)` =>  insert value at specified position  
- `pop(index=-1)`  => removes the value at specified index     

```python
daily_calories.append(2100)
```

```python
daily_calories
```

```python
arr = [10,11,12,13,14,15,16,17]

evens=[]

for num in arr:

    if num%2==0:

        evens.append(num)

print(evens)
```

```python


scores=[45,52,51,43,55,57,65,74,23,24]

passed = []  # >= 50

failed =[] #<50

for mark in scores:

    if mark>=50:

        passed.append(mark)

    else:

        failed.append(mark)

print("passed",passed)

print("failed",failed)
```
