#### write a program to display number is prime or not

```python
def is_prime(number):

    result = True

    for i in range(2,number):

        if number%i==0:

            result=False
            break
    print(result)

is_prime(13)
is_prime(14)
is_prime(17)
```

```python
#write a function for displaying factorial of a number

def factorial(number):

    result = 1

    for i in range(1,number+1):

        result = result * i

    print(result)

factorial(5)
factorial(6)
```

#### perfect number

```python

number = int(input("enter number "))

total = 0

for i in range(1,number):

    if number%i==0:

        total = total + i

print(total==number)



```

```python

def is_perfect_number(number):

    total = 0

    for i in range(1,number):

        if number % i==0:

            total = total + i

    print(total==number)

is_perfect_number(7)
is_perfect_number(28)
```

# harshad number

```python


def is_harsad_number(number):

    number_copy = number

    total = 0

    while(number!=0):

        digit = number % 10

        total = total + digit

        number = number//10

    print(number_copy % total ==0)


is_harsad_number(18)
```
