EG:
```
 word="hello"
```

````python
text="luminar"

pos = text.index("m")

print(pos)
````

`find(value)` => return index position of first occurance of value return -1 if value not exist in str

```
text = "hello"
pos = text.lastindex("l") => 3
```

````python
text="hello"
pos= text.find("v")
print(pos)
````

`rfind(value)` => return index position of last occurance of value return -1 if value not exist
```
    "hello".rfind("l")  => 3
```

````python
text = "hello"

pos=text.rfind("l")

print(pos)
````

`replace(old,new,count)` => replace old_vale with new_value

````python
text="hello"

new_text=text.replace("l","L")

new_text
````

`strip(value)` => remove value from both end

````python
text=",hello,world,"

new_text = text.strip(",")

print(new_text)
````

`lstrip(value)` => remove value from left

````python
text = ",hello"

new_text = text.lstrip(",")

new_text
````

`rstrip(value)` => remove value from right side

````python
text = "hello,"
new_text = text.rstrip(",")
print(new_text)
````

`startswith(prefix)` => return True if the string starts with the specified prefix, False otherwise

````python
text="python programming"

text.startswith("ja")
````

`endswith(suffix)` => Return True if the string ends with the specified suffix, False otherwise

````python
text="python programming"

text.endswith("in")
````

`count(value)` => return frequency of value in string object

````python
text="hello"

frequency = text.count("l")

frequency
````
