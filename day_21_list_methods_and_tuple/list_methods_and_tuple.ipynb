`append(value)`  
`insert(index,value)`  
`pop(index=-1)`  
`remove(value)` remove first occurance of value from list  
`index(value)`  return index of value  
`count(value)`  return frequency of value from list  
`sort()` sort list
`reverse()`  reverse the list in place  


```python
arr = [10,20,30,40,50,20]

print(arr.index(20))

arr.remove(20)

print(arr)
```

```python
arr =[10,11,10,12,10,10,13,14,15]

for num in arr:

    print(num,arr.count(num))
```

```python
arr = [1,4,5,2,7,3,8]

arr.sort(reverse=True)

print(arr)
```

```python
chars = ["t","a","n"]

chars.reverse()

print(chars)
```
