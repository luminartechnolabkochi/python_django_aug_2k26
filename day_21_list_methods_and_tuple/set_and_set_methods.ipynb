#### set
---
`unorderd,unique,mutable`

- define set with `st=set() or st={10,22,23}`

#### set methods
---

- add(value)
- union(sequence)
- intersection(sequence)
- difference(sequence)
- issuperset(sequence)
- issubset(sequence)

```python
st = {10,20,30,40,50,40}

st.add(100)

print(st)
```

```python
st1= {10,20,30,40,50}

st2={10,20,100,200,300,400}

union_set = st1.union(st2)

print(union_set)

intersection_set = st1.intersection(st2)
print(intersection_set)

difference = st2.difference(st1)

print(difference)
```

```python
setA={10,20,30,40,50}
setB={10,20,30}

print(setA.issuperset(setB))

print(setB.issubset(setA))
```

```python

all_students={"hari","sreya","shyam","tom"}

passed={"hari","tom"}

failed= all_students.difference(passed)

print(failed)
```
