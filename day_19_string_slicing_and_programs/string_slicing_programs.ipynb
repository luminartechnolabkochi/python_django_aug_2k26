#### write a function to display vowel count in a string

````python
def vowel_count(text):

    vowles="aeiou"

    v_count = 0

    for ch in text:

        if ch in vowles:

            v_count = v_count+1

    print(v_count)

vowel_count("hello")

vowel_count("pythonprogramming")
````
