```python
#list of dictionary

movies=[
    {"id":1,"title":"kgf","year":2018,"run_time":165,"rating":9},
    {"id":2,"title":"kgf2","year":2023,"run_time":166,"rating":9.5},
    {"id":3,"title":"arm","year":2024,"run_time":160,"rating":8.5},
    {"id":4,"title":"abcd","year":2017,"run_time":156,"rating":8},
    {"id":5,"title":"thudarum","year":2025,"run_time":170,"rating":9.5},
]


#kgf2 runtime
movies[1].get("run_time")

# tudarum year
print("thudarum relaese year",movies[-1].get("year"))

#abcd rating
print("abcd movie rating",movies[-2].get("rating"))
```

```python

all_titles=[di.get("title") for di in movies ]

print(all_titles)
```

```python
#movies released after 2018
year_filter = [di.get("title") for di in movies if di.get("year")>2018]

print(year_filter)
```

```python
#movies released before 2018
year_filter=[di.get("title") for di in movies if di.get("year")<2018]

print(year_filter)
```

```python
#movies with rating > 9

rating_filter = [di.get("title") for di in movies if di.get("rating")>9]

rating_filter
```

```python
# movies with run_time > 158 


run_time_filter=[di.get("title") for di in movies if di.get("run_time")>158]
run_time_filter
```

```python
#movie with highest rating

all_ratings = [di.get("rating") for di in movies ]

max_rating = max(all_ratings)

top_movies=[di.get("title") for di in movies if di.get("rating")==max_rating]

print(top_movies)
```

```python

min_rating = min(all_ratings)

f_movies=[di.get("title") for di in movies if di.get("rating")==min_rating]
f_movies
```

```python

def get_rating(di):
    return di.get("rating")

print(min(movies,key=get_rating))
```
