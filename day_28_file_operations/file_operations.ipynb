```python
#process numbers.txt and print even numbers

fr = open("numbers.txt","r")

for line in fr: #line ="12\n"

    line = line.rstrip("\n") #line = "12"

    num = int(line)

    if num%2==0:

        print(num)
fr.close()
```

```python
#read from words.txt and display palindromes

fr = open("words.txt","r")

for line in fr:

    word = line.rstrip("\n")

    if word==word[::-1]:

        print(word)

fr.close()
```

```python
# display common prefix from words

fr = open("words.txt","r")

all_words = []

for line in fr:

    word = line.rstrip("\n")

    all_words.append(word)

all_words.sort()
first=all_words[0]
last = all_words[-1]

prefix=""

small_word = first if len(first) < len(last) else last

for i in range(len(small_word)):

    if first[i]==last[i]:

        prefix+=last[i]
    else:

        break
print(prefix)

fr.close()
```

```python
#order count

fr = open("orders.txt","r")

all_orders=[]

for line in fr:

    item = line.rstrip("\n")

    all_orders.append(item)


order_count = {o:all_orders.count(o) for o in all_orders}
print(order_count)
```

```python


fr = open("weather.txt","r")

weather_report = []

for line in fr:

    line = line.rstrip("\n")

    district,temp=line.split(",")

    dictionary = {"district":district,"temp":temp}

    weather_report.append(dictionary)

print(weather_report)
```
