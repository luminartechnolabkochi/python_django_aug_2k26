##### Dictionary
----
- `we can define dictionary with ` {}  
- `dictionary store as key:value pairs` 
- `values can be access using key`  

##### Dictionary methods
- `keys()` => return all keys  
- `values()` => all values 
- `items()` => return  both key and value  
- `pop(key)` => remove key value pair   
- `update({key:value,key:value})` 
- `get(key)` => return value by key



```python

expense = [850,750,550,570,750,1000]
#           0    1  2   3   4    5

print(expense[2])
```

```python
expenses={"mon":850,"tue":750,"wed":550,"thu":570,"fri":750,"sat":1000}

print(expenses["tue"])

print(expenses["thu"])
```

```python
calories={"mon":1850,"tue":1780,"wed":2000,"thu":1850,"fri":1900,"sat":2100}


print(calories["fri"])
# adding new key value 
calories["sun"]=2200

calories

#updating 

calories["tue"]+=100

print(calories)
```

```python
# id,title,category,price

item = {"id":10,"title":"lp-shirt","category":"clothings","price":34000}

# display price 

print(item["price"])

# add an offer od 100 rs

item["offer"]=100

print(item)

# update price as current price + 100
item["price"]+=100

print(item)
```

```python
item = {"id":10,"title":"lp-shirt","category":"clothings","price":34000}


for k in item.keys():

    print(k,item[k])
```

```python
for v in item.values():

    print(v)
```

```python
for k,v in item.items():

    print(k,v)
```

```python
item.pop("id")

print(item)
```
